```python
import torch
import torch.nn as nn
import numpy as np
from torch.utils.data import DataLoader
```

```python
# Generate data for AND gate
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[0],[0],[1]])
X_tensor = torch.FloatTensor(X)
y_tensor = torch.FloatTensor(y)
loader = DataLoader(list(zip(X_tensor, y_tensor)), batch_size=1, shuffle=True)
```

```python
# Creating model
class ANDPerceptron(nn.Module):
    def __init__(self):
        super(ANDPerceptron, self).__init__()
        self.linear = nn.Linear(2, 1)  # 2 inputs, 1 output
        self.activation = nn.Sigmoid()  # Activation function

    def forward(self, x):
        x = self.linear(x)
        x = self.activation(x)
        return x
```

```python
# Task 1, Training
model = ANDPerceptron()
n_epochs = 1000
loss = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.1)
model.train()
for epoch in range(n_epochs):
    for batch_X, batch_y in loader:
        y_pred = model(batch_X)
        loss_value = loss(y_pred, batch_y)
        optimizer.zero_grad()
        loss_value.backward()
        optimizer.step()

```

```python
# Task 2, Verification
model.eval()
for i in range(len(X_tensor)):
        y_pred = model(X_tensor[i])
        print(f"Input: {X[i]}, Predicted Output: {y_pred.item():.4f}, Actual Output: {y[i][0]}")
```
